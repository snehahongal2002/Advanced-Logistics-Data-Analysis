# Week 3: Advanced Data Analysis and Visualization in Logistics
This notebook analyzes a simulated logistics dataset.

```
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

df = pd.read_csv('logistics_shipments.csv')
df.head()
```

```
print('Shape:', df.shape)
print('\nMissing values:\n', df.isna().sum())
df.select_dtypes('number').describe().round(2)
```

```
df.groupby('Transport_Mode').agg(
    shipments=('Shipment_ID','count'),
    avg_delivery_days=('Delivery_Time_Days','mean'),
    avg_cost_inr=('Transportation_Cost_INR','mean'),
    avg_cost_per_unit=('Cost_per_Unit_INR','mean')
).round(2)
```

```
df.select_dtypes('number').corr().round(2)
```

```
plt.figure(figsize=(8,5))
plt.hist(df['Delivery_Time_Days'], bins=18, edgecolor='black')
plt.title('Distribution of Delivery Times')
plt.xlabel('Delivery time (days)')
plt.ylabel('Number of shipments')
plt.tight_layout()
plt.show()
```

```
plt.figure(figsize=(8,5))
plt.scatter(df['Distance_km'], df['Transportation_Cost_INR'], alpha=0.7)
plt.xlabel('Distance (km)')
plt.ylabel('Transportation cost (INR)')
plt.title('Distance vs Transportation Cost')
plt.tight_layout()
plt.show()
```

```
df.groupby('Region')['Delivery_Time_Days'].mean().sort_values().plot(kind='bar', edgecolor='black')
plt.ylabel('Average delivery time (days)')
plt.title('Average Delivery Time by Region')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
